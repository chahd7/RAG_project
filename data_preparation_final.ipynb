# Create Meaningful text representations of the data 

```python
#looking again at which columns are present in the dataframe to decide on which ones are important
print(df.columns)
```

For each column, the following decision was taken for the textual representation: 
* Row ID : dropped as it does not provide any meaningful information 
* Order ID: dropped at it does not provide meaningful information 
* Order Date: kept for trend analysis 
* Ship Date : kept for trend analysis 
* Ship Mode: kept as it may provide meaningful information 
* Customer ID: dropped as it does not provide meaningful information since we already have customer name
* Customer Name: kept as it may provide meaningful information 
* Segment: kept as it may provide meaningful information 
* Country: dropped at it only includes one unique value (United States)
* City: kept as it provides meaningful information 
* State: kept as it provides meaningful information 
* Postal Code: dropped as it is not needed since we have City, State and Region already 
* Region: kept as it provides meaningful information 
* Product ID: dropped as it is not needed since we have Category, Sub-Category and Product Name already 
* Category: kept as it provides meaningful information 
* Sub-Category: kept as it provides meaningful information 
* Product Name: kept as it provides meaningful information 
* Sales: kept as it provides meaningful information 
* Discount: kept as it may provide meaningful information 
* Profit: kept as it may  provide meaningful information 


For the natural language representation description for each transaction, I chose to make it follow a logical narrative starting with who then when then what then where and finally how much so that it is well formatted. I also chose to make it into a paragraph of multiple sentences rather than one long sentence to make it more clear to use by the RAG system. The discount was turned into a percentage for a more natural way to interpret it and the profit and sales were rounded up to two decimals to get rid of unecessary noise. 

```python
#create natural language representation description for each transaction 
all_natural_representation = []

for i, row in df.iterrows():

  #seperate them into sentences so that we get rid of the \n and the white spaces in between each sentence. 
  sentences = [
     f"This order was made by the customer {row["Customer Name"]} who belongs to the segment {row["Segment"]}.",
     f"The customer ordered on the date {str(row["Order Date"].date())} and was shipped on the date {str(row["Ship Date"].date())} using the shipment method {row["Ship Mode"]}.",
     f"The customer ordered the product {row["Product Name"]} from the sub-category {row["Sub-Category"]} which belongs to the category {row["Category"]}.",
     f"The customer resides in the city of {row["City"]} which belongs to the state {row["State"]} and to the region {row["Region"]}.",
     f"The product generated {round(row["Sales"], 2)}$ in sales and was ordered with a quantity of {row["Quantity"]} using a discount of {row["Discount"]*100}% and generated a profit of {round(row["Profit"], 2)}$." #rounded so that there are not too many numbers after the decimal which might just confuse the LLM
  ]

  #join into one paragraph
  text = " ".join(sentences)

  #include all metadata relevant to the transaction
  dict_result = {
    "text": text, 
    "customer name": row["Customer Name"],
    "segment": row["Segment"], 
    "order date": str(row["Order Date"].date()), 
    "ship date": str(row["Ship Date"].date()), 
    "category": row["Category"],
    "sub-category": row["Sub-Category"],
    "product name": row["Product Name"], 
    "city": row["City"], 
    "state": row["State"], 
    "region": row["Region"], 
    "sales": round(row["Sales"], 2), 
    "quantity": row["Quantity"],
    "discount": row["Discount"], 
    "profit": round(row["Profit"], 2)
  }

  all_natural_representation.append(dict_result)

#save into json as it is suited for text + metadata and easy to access
with open("json_files2/textual_representation.json", "w", encoding="utf-8") as f:
    json.dump(all_natural_representation, f, indent=4, ensure_ascii=False)

```

# Chunking strategies 

```python
# combine everything into one part 
final_files = [
    "json_files2/textual_representation.json",
    "json_files2/category_info.json",
    "json_files2/city_info.json",
    "json_files2/monthly_sales.json",
    "json_files2/region_info.json",
    "json_files2/state_info.json",
    "json_files2/stats_summary.json",
    "json_files2/subcategory_info.json",
    "json_files2/year_sales.json"
]

combined_data = []
#useee different type files labels accordingly in order to seperate between each type of document, also allows for easier filtering in later steps
types_files = ["transaction", "category_summary", "city_summary", "month_summary", "region_summary", "state_summary", "general_summary", "subcategory_summary", "year_summary"]

for i, file in enumerate(final_files):  
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)

    for d in data: 
        d["type"] = types_files[i]

    combined_data.extend(data)

```

```python
len(combined_data)
```

```python
with open("json_files2/combined.json", "w", encoding="utf-8") as f:
    json.dump(combined_data, f, indent=4, ensure_ascii=False)
```

```python
#chunk based on results
def chunking(text, size): 
    chunks = []
    for i in range(0, len(text), size):
        chunks.append(text[i:i+size])

    return chunks 
```

```python
#gather everything into one place 
def chunk_results(data, size):
    chunk_results = []

    for d in data: 
        chunks = chunking(d["text"], size)

        for i, chunk in enumerate(chunks):
            chunk_df = d.copy()
            chunk_df["id"] = i #add a unique identifier to be used later on in chromaDB
            chunk_df["text"] = chunk
            chunk_results.append(chunk_df)

    return chunk_results      
        
```

```python
#check document length information to help decide on most suited chunking strategies
length_document = [len(c["text"]) for c in combined_data]
print(f"Minimum number characters: {min(length_document)}, Maximum number characters: {max(length_document)}, Average number characters: {sum(length_document)/len(length_document):.0f}")
```

```python
result_500 = chunk_results(combined_data, 500)
result_500
```

It can be noticed here that since 500 is smaller than the average number of characters which is 572, the text gets split weirdly in the middle, making the fixed-size 500 characters approach not suitable 

```python
result_1000 = chunk_results(combined_data, 1000)
result_1000
```

Here since 1000 characters is just slightly above the maximum number of characters, none of the documents get split in the middle and not too much noise is added while also allowing for some flexibility, making it therefore a good potential final chunking strategy option.

```python
result_2000 = chunk_results(combined_data, 2000)
result_2000
```

It can be noticed that the 2000 characters approach gives the exact same results as the 1000 characters approach while also adding more unecessary noise, making it therefore not needed.

```python
#save them 
with open("json_files2/chunks_500.json", "w", encoding="utf-8") as f:
    json.dump(result_500, f, indent=4, ensure_ascii=False)

with open("json_files2/chunks_1000.json", "w", encoding="utf-8") as f:
    json.dump(result_1000, f, indent=4, ensure_ascii=False)

with open("json_files2/chunks_2000.json", "w", encoding="utf-8") as f:
    json.dump(result_2000, f, indent=4, ensure_ascii=False)
```

It can be noticed that with a size of 500, the sentences get cut in the middle. With 1000 it seems to capture the full thing. The result for 2000 returns the same as of 1000 so the additional 1000 are unecessary. Therefore the best solution is 1000. This is also due to the number of average characters we have of 573 and the maximum number of 690. 

# Document-aware Chunking approach

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
#experimenting with different chunk_sizes and chunk_overlap https://lagnchain.readthedocs.io/en/stable/modules/indexes/text_splitters/examples/recursive_text_splitter.html
def split_document(data, chunk_size, chunk_overlap):
    splitter = RecursiveCharacterTextSplitter(
    chunk_size=chunk_size,
    chunk_overlap=chunk_overlap, #allows for some overlap between the chunks to allow for the context to not be cut suddenly
    separators=["\n\n", "\n", ". ", " ", ""] #order of splitting first paragraphs then sentences etc 
    )

    chunked = []
    for d in data: 
        chunks = splitter.split_text(d["text"])

        for i, chunk in enumerate(chunks):
            chunk_df = d.copy()
            chunk_df["id"] = i
            chunk_df["text"] = chunk
            chunked.append(chunk_df)
    return chunked


```

```python
split_document(combined_data, 500, 100)
```

For 500 characters, this method demonstrated better results than fixed chunking as the sentence is no longer cut in the middle but cut in a natural boundary. However the full text is still not included. 

```python
split_document(combined_data, 1000, 100)
```

```python
split_document(combined_data, 2000, 100)
```

While document specific chunking could be a better solution for larger documents, since my documents have an average number of characters of 573, the results for 1000 characters and 2000 characters are the same as obtained with fixed chunking and therefore a fixed-chunking with 1000 characters remains the most ideal and straight-forward solution for this case.
